```python
#Matplot Line Chart(untuk menampilkan pola/tren)

import matplotlib.pyplot as plt

bulan = ["Jan","Feb","Mar","Apr","Mei"]
penjualan = [10,120,150,130,180]

plt.plot(bulan,penjualan)
plt.title("Grafik Penjualan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah Penjualan")
plt.show()
```

```python
#Matplot Bar Chart(Untuk menampilkan )

produk = ["Laptop","Mouse","Keyboard","Monitor"]
terjual = [25,40,30,20]

plt.bar(produk, terjual)

plt.title("Jumlah Produk terjual")
plt.xlabel("Produk")
plt.ylabel("Jumlah Terjual")
plt.show()
```

```python
#Matplot Pie Chart(Untuk menampilkan persentase)

label = ["Python","Java","C++","Javascript"]
jumlah = [45,25,15,15]

plt.pie(jumlah, labels=label, autopct="%1.1f%%")
plt.title("Bahasa Pemrograman Favorit")
plt.show()
```

```python
#Matplot Scatter Plot(Untuk menampilkan hubungan antardua variabel)

jam_belajar = [1,2,3,4,5,6]
nilai = [50,60,68,75,85,95]

plt.scatter(jam_belajar, nilai)

plt.title("Perbandingan Jam Belajar dan Nilai")
plt.xlabel("Jam Belajar")
plt.ylabel("Nilai")
plt.show()
```

```python
#Matplot Histogram(Untuk Menampilkan angka)

nilai = [70,75,80,80,82,85,87,90,90,91,92,95]

plt.hist(nilai)

plt.title("Histogram Nilai")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show()
```

```python
#Matplot Map(untuk menampilkan peta)
import cartopy.crs as ccrs

fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.coastlines()
ax.set_global()
plt.show
```

```python
#Seaborn

import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "Produk":["Laptop","Mouse","Keyboard","Monitor"],
    "Terjual":[50,120,80,40]
})

sns.barplot(data=df,
            x="Produk",
            y="Terjual")

plt.title("Jumlah Produk Terjual")
plt.show
```

```python
#Seaborn Line Chart

df = pd.DataFrame({
"Bulan":("Jan","Feb","Mar","Apr","Mei"),
"Penjualan":[100,120,150,130,180]
})

sns.lineplot(data=df,
             x="Bulan",
             y="Penjualan",
             marker="o")

plt.title("Grafik Penjualan")
plt.show
```

```python
#Seaborn Scatter Plot

df = pd.DataFrame({
    "Jam_Belajar":[1,2,3,4,5,6],
    "Nilai":[50,60,68,75,85,95]
})

sns.scatterplot(data=df,
                x="Jam_Belajar",
                y="Nilai",
                s=100)

plt.title("Perbandingan Jam Belajar vs Nilai")

plt.show()
```

```python
#Seaborn Histogram

nilai = [70,72,74,75,78,80,82,84,85,86,88,90,92,95]

sns.histplot(nilai,
            bins=6,
            kde=True)

plt.title("Distribusi Nilai")

plt.show
```

```python
#Load dataset bawaan Seaborn
tips = sns.load_dataset("tips")

#Membuat Box Plot
sns.boxplot(y=tips["total_bill"])

plt.title("Distribusi Total Tagihan Pelanggan")
plt.ylabel("Total Bill ($)")

plt.show()
```

```python
bulan=["Jan","Feb","Mar","Apr"]
penjualan=[10,120,150,170]

plt.figure(figsize=(8,5))
plt.plot(
               bulan,penjualan,
               color="red",
               marker="o",
               linestyle="--",
               linewidth=3,
               label="Penjualan"
               )

plt.title("Grafik Penjualan")

plt.xlabel("Bulan")

plt.ylabel("Jumlah")

plt.grid(True)

plt.legend()

plt.show
```
