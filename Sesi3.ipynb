DICTIONARY

```python
#Mengakses Data Dictionary
mahasiswa = {
    "nama" : "Andi",
    "umur": 20,
    "prodi":"Informasi"
}
mahasiswa["nama"]

#Menambah Data Disctionary

mahasiswa["ipk"] = 3.85
print(mahasiswa)

mahasiswa["Jenis_Kelamin"] = "Laki_Laki"
print(mahasiswa)

#Mengubah Data Dictionary
mahasiswa["umur"] = 21
print(mahasiswa)

mahasiswa["prodi"] = "Teknik"
print(mahasiswa)

#Mengahapus Data Dictionary
del mahasiswa["umur"]
print(mahasiswa)
```
