```python
def login(user,password):
    return user == "admin" and password == "1234"

def pilih_barang():
    return "Laptop"

def checkout(barang):
    return f"Checkout berhasil untuk {barang}"

#System Test
if login("admin", "1234"):
    barang = pilih_barang()
    hasil = checkout(barang)
    print(hasil)
else:
  print("Login Gagal")
```

Black Box Testing

```python
def login(username, pw):
    data_user ={
        "Andy": "1234",
        "admin": "123"
    }

    if username not in data_user:
        return "User Tidak Ditemui"

    if data_user[username] != pw:
        return "Password Salah"

    return "Login Berhasil"

    assert login("Andy","1234") == "Login Berhasil"
assert login ("pengguna","12345") == "User Tidak Ditemui"
assert login("Andy","12345") == "Password Salah"
print("Semua Acceptance Test Login Berhasil")
```

```python
def cek_kelulusan(nilai):
    if nilai >=75:
        return "Lulus"
    return "Tidak Lulus"

#Black Box Test
print(cek_kelulusan(90))
print(cek_kelulusan(60))
```

White Box Testing

```python
def cek_kelulusan(nilai):
    if nilai >= 75:
        return "Lulus"
    return "Tidak Lulus"

#White Box Test
assert cek_kelulusan(90) == "Lulus"   #Jalur True
assert cek_kelulusan(60) == "Tidak Lulus"   #Jalur False
print("Semua Unit Test Berhasil")
```

```python
#Versi Lama
def login(username, password):
    return username == "admin" and password == "1234"

assert login("admin", "1234") == True

#Mebambahkan fitur baru
def lupa_password(email):
    return "Link reset password setelah dikirim"

#Regression Test
assert login("admin", "1234") == True
assert lupa_password("email") == "Link reset password setelah dikirim"
print("Fitur lama tetap berjalan")
```
