```python
#Menangani dalam kondisi tertentu

try:
  angka=int(input("Masukkan Angka:"))
  hasil=100/angka
except ZeroDivisionError:
  print("NO ZERO")
except ValueError:
  print("Tidak Valid")
```

```python
#Menggunakan else
try:
  angka=int(input("Masukkan Angka:"))
  hasil=100/angka
except ZeroDivisionError:
  print("NO ZERO")
except ValueError:
  print("Tidak Valid")
else:
  print("Hasil=",hasil)
```

```python
#Menggunakan Finally
try:
  angka=int(input("Masukkan Angka:"))
  hasil=100/angka
except ZeroDivisionError:
  print("NO ZERO")
except ValueError:
  print("Tidak Valid")
else:
  print("Hasil=",hasil)
finally:
  print("Program Selesai")
```

```python
#Modul Kalkulator

%%writefile kalkulator.py
def tambah(a,b):
  return a+b
def kurang(a,b):
  return a-b
def kali(a,b):
  return a*b
def bagi(a,b):
  return a/b

```

```python
import kalkulator

print("Penjumlahan:",kalkulator.tambah(10,20))
print("Pengurangan:",kalkulator.kurang(10,20))
print("Perkalian:",kalkulator.kali(10,20))
print("Pembagian:",kalkulator.bagi(10,20))
```

```python
%%writefile bangun_datar.py
pi=3.14
def luas_persegi(s):
  return s*s
def luas_Ppanjang(p,l):
  return p*l
def luas_Segi3(a,t):
  return a*t/2
def luas_lingkaran(r):
  return pi*r*r
```

```python
import bangun_datar as kalkulator

print("luas_persegi:",kalkulator.persegi(10))
print("luas_Ppanjang:",kalkulator.Ppanjang(10,20))
print("luas_Segi3:",kalkulator.Segi3(10,20))
print("Luas_lingkaran:",kalkulator.lingkaran(10))
```

```python
%%writefile nilai.py
def grade(nilai):
  if nilai >= 80:
    return "A"
  elif nilai >= 70:
    return "B"
  elif nilai >= 60:
    return "C"
  elif nilai >= 50:
    return "D"
  else nilai >= 40:
    return "E"
```

```python
import nilai
angka=int(input("Masukkan Angka:"))
print("grade:",nilai.grade(angka))
```
