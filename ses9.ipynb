Inheritance

```python
class Hewan:  #Parent Class

  def makan(self):
      print("Hewan sedang makan")

class Kucing(Hewan):  #Child Class

  def suara(self):
      print("Meong")

k = Kucing()

k.makan()
k.suara()

class Anjing(Hewan): #Child Class

  def makan(self):
      print("Hewan sedang makan")

  def suara(self):
      print("Gung Gung")

a = Anjing()

a.makan()
a.suara()

class Burung(Hewan): #Child Class

    def suara(self):
      print("Cak")

b = Burung()

b.makan()
b.suara()
```

Polymorphins


```python
class Hewan:

  def __init__(self,nama):
      self.nama = nama

class Kucing(Hewan):
  pass

class Anjing(Hewan):
  pass

class Burung(Hewan):
  pass

k = Kucing("Mimi")
a = Anjing("Lulu")
b = Burung("Lili")
print(k.nama)
print(a.nama)
print(b.nama)
```

```python
class Hewan:

  def __init__(self,nama):
      self.nama = nama


class Kucing(Hewan):

  def __init__(self,nama,warna):

    super().__init__(nama)

    self.warna = warna

k = Kucing("lala","coklat")
print(k.nama)
print(k.warna)
```

```python
class Hewan:

  def __init__(self,nama,warna):
      self.nama = nama
      self.warna = warna

class Kucing(Hewan):

  def __init__(self,nama,warna):

    super().__init__(nama,warna)

k = Kucing("lala","coklat")
print(k.nama)
print(k.warna)
```

```python
class Kendaraan:
  def bergerak(self):
      pass

class Mobil(Kendaraan):
  def bergerak(self):
    print("Mobil berjalan dengan roda")

class Pesawat(Kendaraan):
  def bergerak(self):
    print("Pesawat terbang di udara")

class Kapal(Kendaraan):
    def bergerak(self):
        print("Kapal berlayar di laut")

kendaraan = [
    Mobil(),
    Pesawat(),
    Kapal()
]

for k in kendaraan:
    k.bergerak()
```

```python
class Pembayaran:
  def bayar(self,jumlah):
      pass

class TransferBank(Pembayaran):
  def bayar(self,jumlah):
      print(f"TransferBank sebesar Rp{jumlah:,}")

class QRIS(Pembayaran):
    def bayar(self,jumlah):
      print(f"Scan QRIS sebesar Rp{jumlah:,}")

class KartuKredit(Pembayaran):
  def bayar(self,jumlah):
      print(f"Kartu Kredit sebesar Rp{jumlah:,}")

metode = [
    TransferBank(),
    QRIS(),
    KartuKredit()
]

for m in metode:
    m.bayar(50000)
```

```python
class Character:
  def attack(self):
      pass

class Warrior(Character):
  def attack(self):
    print("Warrior menyerang dengan pedang")

class Archer(Character):
  def attack(self):
    print("Archer membak Panah")

class Mage(Character):
  def attack(self):
    print("Mage mengeluarkan bola api")

tim = [
    Warrior(),
    Archer(),
    Mage()
]

for hero in tim:
    hero.attack()
```

```python
def faktorial(n):
    if n == 1:
        return 1

    return n * faktorial(n - 1)

print(faktorial(3))
```

```python
def jumlah(n):
    if n ==1:
        return 1

    return n + jumlah(n - 1)

print(jumlah(5))
```

```python
def balik(teks):
    if len(teks) ==0:
        return ""

    return balik(teks[1:]) + teks[0]

print(balik("harvey"))
```
